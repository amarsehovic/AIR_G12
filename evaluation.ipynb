```python
# Import necessary libraries
import matplotlib.pyplot as plt
import numpy as np

# Evaluation Metrics Data
# Random Forest Metrics
rf_metrics = {
    "Precision": 0.6449,
    "Recall": 0.6414,
    "F1-Score": 0.6410,
    "Accuracy": 0.64
}

# Collaborative Filtering Metrics
cf_metrics = {
    "Precision": 0.6688,
    "Recall": 0.5957,
    "F1-Score": 0.5333,
    "Accuracy": 0.60
}

# Create bar plot to compare metrics
categories = list(rf_metrics.keys())
rf_values = list(rf_metrics.values())
cf_values = list(cf_metrics.values())

x = np.arange(len(categories))  # Label locations
width = 0.35  # Bar width

fig, ax = plt.subplots(figsize=(10, 6))
bars_rf = ax.bar(x - width / 2, rf_values, width, label="Random Forest")
bars_cf = ax.bar(x + width / 2, cf_values, width, label="Collaborative Filtering")

# Add labels, title, and legend
ax.set_xlabel("Metrics")
ax.set_ylabel("Scores")
ax.set_title("Comparison of Evaluation Metrics")
ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.legend()

# Add values on top of bars
for bars in [bars_rf, bars_cf]:
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height:.2f}', xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 3), textcoords="offset points", ha='center', va='bottom')

plt.tight_layout()
plt.show()
```
